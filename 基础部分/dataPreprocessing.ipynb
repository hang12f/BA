```python
import pandas as pd
import re
```

# 数据集的导入和基本信息的查询

```python
# line.decode("utf-8","ignore")
# 读取数据
dataPath = './data/BA_AirlineReviews.csv'
data = pd.read_csv(dataPath, encoding='latin-1')

data.head(5)
```

```python
# 查看数据集大小
data.shape
```

# 数据清洗

## 删除无关特征

```python
# 查看数据的列名
columns = data.columns.tolist()
print(columns)
```

```python
# 删除一些本次作业用不上的列
columns_to_drop = ['Unnamed: 0', 'ReviewBody', 'Name', 'Datetime', 'VerifiedReview', 'ValueForMoney', 'Recommended']
data = data.drop(columns=columns_to_drop)
data.head()
```

```python
# 查看数据集大小
data.shape
```

## 处理空缺值

```python
# 计算数据每列的空缺率
missing_values = data.isna().sum() / len(data)
print(missing_values)
```

```python
# 再删除一些空缺率高的列
columns_to_drop = ['InflightEntertainment', 'Wifi&Connectivity']
data = data.drop(columns=columns_to_drop)
data.head()
```

```python
# 删除含有空缺值的数据
data.dropna(inplace=True)
```

```python
# 检查每列的空缺率
missing_values = data.isna().sum() / len(data)
print(missing_values)
```

```python
data.shape
```

## 文本规范化

```python
# 纠正一些机型中存在错误
data['Aircraft'] = data['Aircraft'].replace('Boeing 747- 400', 'Boeing 747-400')
data['Aircraft'] = data['Aircraft'].replace('Boeing 747 400', 'Boeing 747-400')
data['Aircraft'] = data['Aircraft'].replace('E-170', 'E170')
data['Aircraft'] = data['Aircraft'].replace('E-190', 'E190')
data['Aircraft'] = data['Aircraft'].replace('Embraer170', 'Embraer 170')
data['Aircraft'] = data['Aircraft'].replace('Embraer-190', 'Embraer 190')
```

## 增加新字段

```python
# 提取文本中的月份
data['DateFlown'] = data['DateFlown'].str.split('-').str[1]
data['DateFlown']
```

```python
# 定义一个函数来提取城市名称并拆分成两部分
def split_route(route):
    match = re.match(r'(.+) to (.+)', route)
    if match:
        return match.group(1), match.group(2)
    else:
        return None, None

# 创建两个新列
data['Departure'], data['Destination'] = zip(*data['Route'].apply(split_route))

# 删除原始的 Route 列
data.drop(columns=['Route'], inplace=True)
```

```python
# 定义要替换的字典，键是要被替换的值，值是替换后的新值
replace_dict = {"LHR": "Heathrow", "London Heathrow": "Heathrow", "LGW": "Gatwick", "London Gatwick": "Gatwick", 
                "BKK": "Suvarnabhumi", "MAN": "Manchester", "JFK": "John F. Kennedy", "GLA": "Glasgow", 
                "LCY": "London", "SIN": "Singapore", "YYZ": "Toronto", "ATH": "Athens", "MIA": "Miami", 
                "JNB": "O.R. Tambo", "SFO": "San Francisco", "AMS": "Amsterdam", "YVR": "Vancouver", "HKG": "Hong Kong", 
                "London ": "London", "CPT": "Cape Town", "KUL": "Kuala Lumpur", "GVA": "Geneva", "DXB": "Dubai" , 
                "BOS": "Boston", "EDI": "Edinburgh", "BGI": "Grantley Adams"}
# 使用循环遍历字典中的每一对键值对，使用when函数进行条件判断，并使用withColumn方法进行替换
for old_value, new_value in replace_dict.items():
    data['Departure'] = data['Departure'].replace(old_value, new_value)
    data['Destination'] = data['Destination'].replace(old_value, new_value)
```
